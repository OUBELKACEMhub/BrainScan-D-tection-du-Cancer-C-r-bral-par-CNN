```python
import os
import cv2
import numpy as np

def load_and_filter_images(dataset_path, img_size=(224, 224)):
    """
    Parcourt les dossiers de classes, vérifie les extensions valides,
    nettoie les fichiers non conformes, redimensionne et normalise les images.
    """
    # 1. Définir les extensions autorisées
    valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
    
    images = []
    labels = []
    
    # Vérifier si le dossier principal existe
    if not os.path.exists(dataset_path):
        raise FileNotFoundError(f"Le dossier {dataset_path} est introuvable.")
        
    # 2. Parcourir les dossiers de premier niveau (chaque dossier représente une classe)
    classes = sorted(os.listdir(dataset_path))
    print(f"Classes détectées : {classes}")
    
    for class_name in classes:
        class_folder = os.path.join(dataset_path, class_name)
        
        # S'assurer que c'est bien un dossier
        if not os.path.isdir(class_folder):
            continue
            
        print(f"Analyse du dossier de la classe : {class_name}")
        
        # 3. Parcourir les fichiers à l'intérieur du dossier de la classe
        for filename in os.listdir(class_folder):
            file_path = os.path.join(class_folder, filename)
            
            # Vérifier si c'est un fichier (ignorer les sous-dossiers éventuels)
            if not os.path.isfile(file_path):
                continue
                
            # 4. Vérifier l'extension du fichier (insensible à la casse .JPG ou .jpg)
            if not filename.lower().endswith(valid_extensions):
                print(f"-> Fichier ignoré (mauvaise extension) : {filename}")
                continue
                
            # 5. Charger l'image avec un bloc try-except pour gérer les fichiers corrompus
            try:
                img = cv2.imread(file_path)
                
                # Vérifier si OpenCV a bien pu lire l'image
                if img is None:
                    print(f"-> Fichier corrompu ou illisible : {filename}")
                    continue
                    
                # Redimensionnement de l'image (ex: 224x224)
                img_resized = cv2.resize(img, img_size)
                
                # Ajout aux listes
                images.append(img_resized)
                labels.append(class_name)
                
            except Exception as e:
                print(f"-> Erreur lors du traitement de {filename}: {e}")
                
    # Conversion en tableaux NumPy et normalisation des pixels [0, 1]
    X = np.array(images, dtype=np.float32) / 255.0
    y = np.array(labels)
    
    print(f"\nChargement terminé avec succès !")
    print(f"Nombre total d'images valides chargées : {X.shape[0]}")
    
    return X, y, classes
```
